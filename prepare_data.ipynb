```python
import json
import yaml
import os
import re
```

```python
# Read data
train_data = []
with open('data/train', 'r', encoding='utf-8') as f:
    for line in f:
        line = line.strip()
        if line:
            train_data.append(json.loads(line))
```

```python
print(len(train_data))
```

```python
# Filter length
clean_data_length = []
bad_data_length = []
for sample in train_data:
    len_src = len(sample['src'])
    len_dst = len(sample['dst'])
    if ((len_src/len_dst > 2) or (len_src/len_dst < 1/2)):
        bad_data_length.append(sample)
    else:
        clean_data_length.append(sample)
```

```python
len(clean_data_length)
```

```python
# Filter punctuation
clean_data_punct = []
bad_data_punct = []
for sample in clean_data_length:  
    if(set(re.findall(r'[!?]', sample['src']))==set(re.findall(r'[!?]', sample['dst']))):      
        clean_data_punct.append(sample)
    else:
        bad_data_punct.append(sample)
```

```python
with open('data/train_clean', 'w', encoding='utf-8') as f:
    for item in clean_data_punct:
        json_line = json.dumps(item, ensure_ascii=False)
        f.write(json_line + '\n')
```
